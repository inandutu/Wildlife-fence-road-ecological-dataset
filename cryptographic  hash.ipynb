```python
import hashlib
from pathlib import Path
import os
```

```python
#directory path
dataset_path = Path("/Users/triscillar/Documents/Professional development/Journal Submissions/data MDPI/Wildlife Fences Dataset/Still_Images/Segmentation/semantic level masks/images")
```

```python
splits = ["train_split", "val_split", "test_split"]
image_extensions = {".png"}
```

```python
#COLLECT ALL IMAGES
images = []

for split in splits:
    split_dir = dataset_path / split

    if not split_dir.exists():
        print(f"WARNING: {split_dir} does not exist")
        continue

    for path in split_dir.rglob("*"):
        if path.is_file() and path.suffix.lower() in image_extensions:
            images.append({
                "path": path,
                "split": split
            })

print(f"Total images found: {len(images)}")

for split in splits:
    count = sum(1 for x in images if x["split"] == split)
    print(f"{split}: {count}")
```

```python
def calculate_image_hash(image_path):
    """Calculate SHA-256 hash of the image file."""

    hash_obj = hashlib.sha256()

    with open(image_path, "rb") as image_file:
        for chunk in iter(lambda: image_file.read(8192), b""):
            hash_obj.update(chunk)

    return hash_obj.hexdigest()
```

```python
def find_similar_images(dataset_path):
    """Find exact duplicate images in all subfolders."""
    image_hashes = {}
    total_images = 0
    duplicate_count = 0

    # Walk through all folders and subfolders
    for root, folders, files in os.walk(dataset_path):
        for image_file in files:
            if image_file.lower().endswith(
                ('.png')
            ):
                image_path = os.path.join(root, image_file)
                total_images += 1

                # Calculate SHA-256 hash
                image_hash = calculate_image_hash(image_path)
                # Check whether this hash already exists
                if image_hash in image_hashes:
                    duplicate_count += 1
                    
                    print("\nDuplicate found:")
                    print("Image 1:", image_hashes[image_hash])
                    print("Image 2:", image_path)
                    print("SHA-256:", image_hash)

                else:
                    image_hashes[image_hash] = image_path

    print("\n===================================")
    print("SHA-256 DUPLICATE ANALYSIS")
    print("===================================")
    print("Total images analysed:", total_images)
    print("Unique images:", len(image_hashes))
    print("Exact duplicates:", duplicate_count)
    print("===================================")
```

```python
find_similar_images(dataset_path)
```
